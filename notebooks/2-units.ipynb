## Michelson data example - compute the displacement distance

Only for the first data set from the first row.

```python
data = [257.36   ,  28.672  , 114.55   ,   0.99614]
```

```python
## metadata 
#
N = data[0] / u.s            # revolutions per second
r = data[1] * u.imperial.ft  # feet
turns = data[2]              # dimensionless: turns of the screw
pitch = data[3] * u.mm       # millimetres per turn
```

```python
# compute displacement d = turns * pitch
d = turns * pitch
d
```

```python
# Compute the speed of light with units
```

```python
import math

# distance to other mirror
D = 1986.23 * u.imperial.ft

# equation for speed of light c
c = 8 * math.pi * N * D * r / d
c
```

```python
# request desired units
c.to("m/s")
```
