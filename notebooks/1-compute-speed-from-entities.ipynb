# 1. Re-compute the speed of light from Michelson's instrument readings

```python
import mammos_entity as me
```

```python
data = me.from_csv("data/michelson-1879-entities.csv")
```

```python
data
```

```python
data.N  # expand by clickling on [+] - only in live version of notebook
```

```python
# explore data object - can get to the underlying quantity (numpy array with units)
data.N.quantity
```

```python
# Constant D defined in free text description above

import mammos_units as u
D = 605.4029 * u.m
```

```python
# we need pi
import math
pi = math.pi
```

```python
# extract (quantity) data from data set 

N = data.N.quantity  # numpy array with units
r = data.r.q  # .q is abbreviation for .quantity
d = data.d.q
# compute speed of light using equation from metadata
c = 8 * pi * N * D * r / d
c
```

```python
# We want the mean:
c.mean()
```

```python
# in m/s
c.mean().to("m/s")
# export to pandas
```

```python
# or km/s
c.mean().to("km/s")
```
