# Entity collection example

Create 2 entities and save them together in a file.

```python
import mammos_entity as me
import mammos_units as u
```

## 1. Creating entities for cable length and cable mass

```python
# We want to record the length of 3 cables.

# 1. Start with a quantity for three different values
L_q = [10, 25, 42] * u.m
```

```python
L_q
```

```python
# Entities can also carry arrays as values
L = me.Entity("Length", [10, 25, 42], "m", 
              description="length of cable")
L
```

```python
# imagine we know how heavy each metre of a cable is.
# We call the cable per unit length density "rho"
```

```python
rho = me.Entity("LinearMassDensity", 2.5, u.kg/u.m, 
                description="mass of the cable per running metre")
rho
```

```python
rho.q * L.q   # compute mass for each cable
```

```python
# turn mass into entity
m = me.Entity("Mass", rho.q * L.q, 
              description="mass of the cable - must not exceed 250kg")  
m
```

## 2. Create entity collection and save to disk

```python
data = me.EntityCollection("Undersea cable",
                           L=L, 
                           m=m)
data
```

```python
data.to_yaml("/tmp/data2.yaml")
```

```python
data.to_csv("/tmp/data2.csv")
```
