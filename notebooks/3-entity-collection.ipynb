# Entity collection example

- Create 2 entities
- combine them in an *entity collection*
- save the entity collection in a file

```python
import mammos_entity as me
import mammos_units as u
```

## 1. Creating entities for cable length and cable mass

```python
# We want to record the length and mass of 3 (thick) cables.

# 1. Start with a quantity for three different values
L_q = [10, 25, 42] * u.m
```

```python
L_q
```

```python
# Entities can also carry arrays as values
L = me.Entity("Length", [10, 25, 42], "m", 
              description="length of cable")
L
```

```python
# We know the cable has a linear mass density of 2.5 kg / m. 
rho = 2.5 * u.kg/u.m 
```

```python
rho * L.q   # compute mass for each cable
```

```python
# turn mass array into entity
m = me.Entity("Mass", rho * L.q, 
              description="mass of the cable - must not exceed 250kg")  
m
```

## 2. Create entity collection and save to disk

```python
data = me.EntityCollection("Undersea cable",
                           L=L, 
                           m=m)
data
```

```python
data.L
```

```python
data.to_yaml("cable-data.yaml")
```

```python
data.to_csv("cable-data.csv")
```

```python
data.to_hdf5("cable-data.h5")
```

```python
# How to read the data back in from disk?
data2 = me.from_yaml("cable-data.yaml")
data2
```
