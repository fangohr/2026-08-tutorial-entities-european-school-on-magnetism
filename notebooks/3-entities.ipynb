## Introduction to (mammos-) entities


```python
import mammos_entity as me
import mammos_units as u  # very much the same as astropy.units
```

```python
T0 = 300
```

```python
# add units (Kelvin)
T1 = 300 * u.K
T1
```

```python
# create entity with ontology label (ThermodynamicTemperature)
T2 = me.Entity("ThermodynamicTemperature", 300, "K")
T2
```

```python
# create entity with description "Used as reference for room temperature"
T2 = me.Entity("ThermodynamicTemperature", 300, "K", description="room temperature reference")
T2
```

```python
# how do I know what ontology labels exist?
me.search_labels("Temperature")
```

```python
# Practical problem: Attempt calculation with entity
2. * T2
```

Calculations with entities are not possible (because the resulting object is likely to need a different and unknown ontology label).

Instead, use the `quantity` part of the entity via `.quantity` or `.q`:

```python
2 * T2.q
```
